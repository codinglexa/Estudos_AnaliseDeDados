```python
# Importando as bibliotecas

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Importando os dados

df = pd.read_csv('/content/lol-jungler-ranked-euw-2026.csv')
```

```python
# Renomeando as colunas

df.rename(columns={
    'rank': 'Ranque',
    'champion': 'Campeão',
    'win': 'Vitória',
    'kills': 'Kills',
    'deaths': 'Mortes',
    'assists': 'Assistências',
    'kp': 'Participações',
    'cs': 'Farm',
    'cspm': 'Farm por Minuto',
    'gold': 'Ouro',
    'dmg': 'Dano',
    'dmgshare': '% Dano',
    'dmgtaken': 'Dano sofrido',
    'vision': 'Visão',
    'wards': 'Wards',
    'wardskilled': 'Wards Destruídas',
    'cwards': 'Wards de Controle',
    'teamdragons': 'Dragões',
    'teambarons': 'Barons',
    'teamheralds': 'Arautos',
    'teamtowers': 'Torres',
    'pdrag': 'Participação em Dragões',
    'pbaron': 'Participação em Barons',
    'pturret': 'Participação em Torres',
    'keystone': 'Runa',
    'dur': 'Duração da Partida',
    'matchId': 'ID da Partida'
}, inplace=True)
```

```python
# Criando a coluna KDA e colocando na quarta posição (índice 3)

df.insert(
    3,
    "KDA",
    np.where(
        df["Mortes"] == 0,
        df["Kills"] + df["Assistências"],
        (df["Kills"] + df["Assistências"]) / df["Mortes"]
    )
)
```

```python
# Deletando a linha onde Ranque = Master (1 único registro)
df = df[df["Ranque"] != "MASTER"]
```

```python
# Padronizando os campos de ["Vitória"]

df["Vitória_num"] = df["Vitória"].map({
    "1": 1,
    "0": 0,
    "true": 1,
    "false": 0
})
```

```python
df.info()
```

## ⚡ Análise de métricas por ranque

```python
# Quantidade de jogadores por ranque

df['Ranque'].value_counts()
```

```python
# Média dos objetivos e métricas por ranque

df.groupby("Ranque")[["Dragões", "Barons", "Arautos", "Torres", "Wards", "Wards de Controle", "Duração da Partida"]].mean().round(2)
```

## 🗼 Correlação de objetivos com vitória

```python
# As métricas quando o jogador vence

df[df["Vitória_num"] == 1].groupby("Ranque")[["Dragões", "Barons", "Arautos", "Torres", "Wards", "Wards de Controle", "Duração da Partida"]].mean().round(2)
```

```python
# As métricas quando o jogador perde

df[df["Vitória_num"] == 0].groupby("Ranque")[["Dragões", "Barons", "Arautos", "Torres", "Wards", "Wards de Controle", "Duração da Partida"]].mean().round(2)
```

## ⛔ Importância de um bom KDA

```python
# Média de KDA nas vitórias

df[df["Vitória_num"] == 1].groupby("Ranque")[["KDA"]].mean().round(2)
```

```python
# Média de KDA nas derrotas

df[df["Vitória_num"] == 0].groupby("Ranque")[["KDA"]].mean().round(2)
```

## 🥇 Importância de farm e ouro

```python
# Média de farm nas derrotas

df[df["Vitória_num"] == 0].groupby("Ranque")[["Farm"]].mean().round(2)
```

```python
# Média de ouro nas vitórias

df[df["Vitória_num"] == 1].groupby("Ranque")[["Ouro"]].mean().round(2)
```

```python
# Média de ouro nas derrotas

df[df["Vitória_num"] == 0].groupby("Ranque")[["Ouro"]].mean().round(2)
```

## 👀 Diferença do uso de visão por ranque

```python
# Média de visão

df.groupby("Ranque")[["Visão", "Wards", "Wards Destruídas", "Wards de Controle"]].mean().round(2)
```

```python
# Média de visão nas vitórias

df[df["Vitória_num"] == 1].groupby("Ranque")[["Visão", "Wards", "Wards Destruídas", "Wards de Controle"]].mean().round(2)
```

```python
# Média de visão nas derrotas

df[df["Vitória_num"] == 0].groupby("Ranque")[["Visão", "Wards", "Wards Destruídas", "Wards de Controle"]].mean().round(2)
```

## 🌴 Campeão com mais vitórias e derrotas

```python
# Campeões com mais vitórias

df[df["Vitória_num"] == 1]["Campeão"].value_counts().head()
```

```python
# Taxa de vitória por elo (10 ou mais jogos)

resultado = (
    df.groupby(["Ranque", "Campeão"])
      .agg(
          Jogos=("Vitória_num", "count"),
          Taxa_Vitoria=("Vitória_num", "mean")
      )
      .query("Jogos > 9")
)

resultado["Taxa_Vitoria"] = (resultado["Taxa_Vitoria"] * 100).round(2)

resultado.reset_index().groupby("Ranque").head(1)
```

```python
# Campeões com mais derrotas

df[df["Vitória_num"] == 0]["Campeão"].value_counts().head()
```

## 🧿 Runas escolhidas

```python
# Runas escolhidas

df['Runa'].value_counts()
```

```python
# Campeões que mais utilizaram a runa Conqueror

df[df["Runa"] == "Conqueror"]["Campeão"].value_counts().head(10)
```

```python
# Kayn sendo o campeão que mais venceu, quais runas foram escolhidas nas vitórias

df[(df["Campeão"] == "Kayn") & (df["Vitória"] == "1")]["Runa"].value_counts()
```

## ☯ Correlações

```python
# Correlação entre ouro e dano

df["Dano"].corr(df["Ouro"])
```

```python
# Correlações entre diversas métricas e a vitória

df.groupby("Ranque").apply(
    lambda grupo: grupo[["Vitória_num","KDA", "Dano", "Farm", "Ouro", "Dragões", "Barons", "Arautos", "Visão"]]
    .corr()["Vitória_num"]
)
```
